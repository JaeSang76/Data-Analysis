```python
import numpy as np

# 스칼라 (0차원센터) : 하나의 숫자
s = np.array(18.2)
print(f"\n np.array(18.2) 스칼라 차원 : {s.ndim}, 모양 : {s.shape}")

# 벡터(1차원 텐서) : 숫자가 한 줄로 나열된 형
v = np.array([92.0, 85.5, 78.2])
print(f"\n v = np.array([92.0, 85.5, 78.2]) : {v}")
print(f"\n 인스턴스 v의 몇차원 텐서 v.ndim: {v.ndim}")
print(f"\n 인스턴스 v의 벡터 차원 v.shape: {v.shape}")
print(f"\n 인스턴스 v의 평균벡터 v.mean(): {v.mean()}")
print(f"\n 인스턴스 v의 모분산 v.var(): {v.var()}")

# 벡터(2차원 텐서) : 행과 열로 구성된 표 형태
m = np.array([[90,80], [75,88]]) # 2X2행렬  [90, 80]
                                  #         [75, 88]
print(f"\n m = np.array([90,80], [75, 88]) : {m}")
print(f"\n 인스턴스 m의 몇차원 텐서 v.ndim: {m.ndim}")
print(f"\n 인스턴스 m의 벡터 차원 v.shape: {m.shape}")
print(f"\n 인스턴스 m의 평균벡터 v.mean(): {m.mean()}")
print(f"\n 인스턴스 m의 모분산 v.var(): {m.var()}")
```

```python
import pandas as pd
import numpy as np

# 1. 기초 데이터(3명의 학생의 2개 과목 점수)
np_matrix = np.array([
                        [90, 80],
                        [75, 88],
                        [92, 95]
                    ])

# 2. NumPy 배열을 Pandas DataFrame형식으로 변환
df = pd.DataFrame(np_matrix,
                  columns = ['통계학', 'P/G'],
                  index = ['A', 'B', 'C'] )
print(f"\n NumPy 배열을 Pandas DataFrame형식으로 변환: \n {df}")
print(f"\n NumPy 배열의 값 : \n {df.values}")
```

```python
# 벡터와 행렬의 연산
import numpy as np

A = np.array([[1,2], [3,4]])
B = np.array([[1,1], [0,1]])

# 덧셈, 뺄셈 및 스칼라배
print(f"행렬의 덧셈/뺄셈 : A+B \n {A+B}")
print(f"행렬의 스칼라배 : 3 * A \n {3 * A}")
```

```python
# 벡터와 행렬의 연산
import numpy as np

v1 = np.array([1,2])
v2 = np.array([4,6])

# 벡터의 내적(dot product)
dot_res = np.dot(v1, v2)
print(f"벡터의 내적 : np.dot(v1, v2) \n {dot_res}")
print(f"벡터의 내적 : v1.T @ v2 \n {v1.T @ v2}")
print(f"벡터의 내적 : v1 @ v2 \n {v1 @ v2}")
```

```python
# 벡터와 행렬의 연산
import numpy as np

A = np.array([[1,2], [3,4]])
B = np.array([[1,1], [0,1]])

# 행렬의 곱(matrix multiplication)
matmul_res = np.matmul(A, B)
print(f"행렬의 곱 : np.matmul(A, B) \n {matmul_res}")
print(f"행렬의 곱 : A @ B \n {A @ B}")
print(f"원소별 곲 : A * B \n {A * B}")
```

```python
# 벡터와 행렬의 구조 변화
import numpy as np

# 4개의 원소를 가진 1차원 벡터를 2X2 행렬로 변환
v1 = np.array([92.0, 85.5, 78.2, 10.0])
mat_2X2 = v1.reshape(2,2) # v1.reshape(-1,2)으로도 가능

print(f"2X2 행렬로 변환 : v1.reshape(2,2) \n {mat_2X2}")
print(f"2X2 행렬로 변환 : v1.reshape(-1,2) \n {v1.reshape(-1,2)}")
print(f"2X2 행렬로 변환 : mat_2X2.shape \n {mat_2X2.shape}")

# 4개의 원소를 가진 1차원 벡터를 열벡터로 변환
mat_4X1 = v1.reshape(4, 1) # v1.reshape(-1,1)으로도 가능

print(f"4X1 행렬로 변환 : v1.reshape(4, 1) \n {mat_4X1}")
print(f"4X1 행렬로 변환 : mat_4X1.shape \n {mat_4X1.shape}")

print(f"1x4 행렬로 전치변환 : mat_4X1.T \n {mat_4X1.T}")
print(f"1x4 행렬로 전치변환 : mat_4X1.T \n {mat_4X1.T.shape}")
```
