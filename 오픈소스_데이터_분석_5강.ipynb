## 5-2 JSON 형식 저장

###orient 옵션 (JSON의 구조적 형태 결정)
'records' : 행(Row) 하나하나를 딕셔너리로 만들고, 이를 리스트로 묶습니다. 웹 API나 대다수 NoSQL DB 표준 형식입니다.

'columns' : 열(Column) 이름을 가장 바깥쪽 키로 잡고, 내부에 {인덱스: 값} 형태로 채웁니다.

'index' : 행 인덱스를 가장 바깥쪽 키로 잡고, 내부에 {열이름: 값} 형태로 채웁니다.

'values' : 열 이름이나 인덱스를 모두 버리고, 순수한 데이터 값만 2차원 배열 형태로 저장합니다. 용량이 가장 작습니다.

```python
### 라이브러리 임포트
import json
import pandas as pd

data = {
    "이름": "홍길동",
    "나이": 25,
    "거주지": "서울",
    "관심사": ["프로그래밍", "데이터 분석", "여행"]
}
df = pd.DataFrame(data)
## json.dump를 이용한 json 저장
with open('data_dump.json', 'w', encoding='utf-8') as f:
    json.dump(df.to_dict(orient='records'), f, ensure_ascii=False, indent=4)
## DataFrame을 이용한 json 저장
df.to_json('data_pandas.json', orient='records', force_ascii=False, indent=4)
```

## 5-3 Pandas Dataframe 데이터 추출

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터학", "경영학", "농학", "교육학", "영어영문학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}
df = pd.DataFrame([data])

## DataFrame 인덱스 출력
print('index :\n', df.index)
## DataFrame 컬럼 출력
print('\ncolumns :\n', df.columns)
## DataFrame 행 출력
print('\nrows :\n', df.values.tolist())
## DataFrame 값 출력
print('\nvalues :\n', df.values)
```

## 5-4 DataFrame 생성 방법

```python
import io
import json
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}

## 딕셔너리, csv, json으로 부터 DataFrame 생성

# 1. 딕셔너리(Dictionary)로부터 생성
df_from_dict = pd.DataFrame(data)

# 임시 테스트를 위한 CSV 및 JSON 데이터 생성 (문자열 형태)
csv_data = """이름,학년,학점,학과,동아리
김철수,1,4.2,컴퓨터공학,프로그래밍
이영희,2,3.8,경영학,독서토론
박민수,3,4.5,전자공학,로봇공학
최지훈,4,3.9,의학,봉사활동
정소희,2,3.5,심리학,음악감상"""

json_data = json.dumps(data, ensure_ascii=False)

# 2. CSV 문자열로부터 생성 (io.StringIO로 파일처럼 읽기)
df_from_csv = pd.read_csv(io.StringIO(csv_data))

# 3. JSON 문자열로부터 생성
df_from_json = pd.read_json(io.StringIO(json_data))


## 세가지 방법 결과 비교
print("1. 딕셔너리로 만든 DataFrame:")
print(df_from_dict.head(2))  # 상위 2개 행만 출력

print("\n2. CSV 문자열로 만든 DataFrame:")
print(df_from_csv.head(2))

print("\n3. JSON 문자열로 만든 DataFrame:")
print(df_from_json.head(2))

# 세 데이터프레임이 완전히 동일한지 확인하는 로직
# 같으면 True, 다르면 False 반환
is_equal_dict_csv = df_from_dict.equals(df_from_csv)
is_equal_dict_json = df_from_dict.equals(df_from_json)

print(f"\n[비교 결과] 딕셔너리 vs CSV 결과가 같은가요?: {is_equal_dict_csv}")
print(f"[비교 결과] 딕셔너리 vs JSON 결과가 같은가요?: {is_equal_dict_json}")
```

## 5-5 DataFrame 저장 방법

```python
import pandas as pd

data = {
    "이름": ["김철수", "이영희", "박민수", "최지훈", "정소희"],
    "학년": [1, 2, 3, 4, 2],
    "학점": [4.2, 3.8, 4.5, 3.9, 3.5],
    "학과": ["컴퓨터공학", "경영학", "전자공학", "의학", "심리학"],
    "동아리": ["프로그래밍", "독서토론", "로봇공학", "봉사활동", "음악감상"]
}

## Dataframe 생성
df = pd.DataFrame(data)
## CSV, JSON 형식으로 저장
print('CSV, JSON 형식으로 저장\n')
df.to_csv('students.csv', index=False, encoding='utf-8-sig')
print('\nJSON 형식으로 저장\n')
df.to_json('students.json', orient='records', force_ascii=False, indent=4)
## HTML 형식으로 저장
print('\nhtml 형식으로 저장\n')
df.to_html('students.html', index=False, encoding='utf-8')
```
