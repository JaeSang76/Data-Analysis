## 3-1 언패킹

```python
# 튜플 언패킹
coordinates = (10, 20)
x, y = coordinates
print(f'x좌표: {x}, y좌표: {y}')

# 리스트 언패킹
rgb = [255, 128, 0]
red, green, blue = rgb
print(f'빨강: {red}, 초록: {green}, 파랑: {blue}')

# 문자열 언패킹
word = 'ABC'
first, second, third = word
print(f'첫 번째 문자: {first}, 두 번째 문자: {second}, 세 번째 문자: {third}')
```

```python
# 함수값 return 언패킹
def get_dataset_stats(values):
  return min(values), max(values), sum(values) / len(values)

data = [12, 8, 21, 17, 5, 32, 14]

minimum, maximum, average = get_dataset_stats(data)
print(f'최솟값: {minimum}, 최댓값: {maximum}, 평균: {average:.2f}')
```

```python
# 반복문에서 언패킹 사용
student_scores = [
    ('김철수', 85, 92, 78),
    ('이영희', 92, 88, 95),
    ('박지민', 75, 83, 90)
]

for name, database, python, cloud in student_scores :
  average = (database + python + cloud) / 3
  print(f'{name}의 평균 점수: {average:.1f}')

runners = ['철수', '영희', '민수']
for rank, name in enumerate(runners, start=1):
    print(f'{rank}등: {name}')

# 리스트 컴프리헨션 활용
numbers = [1, 2, 3, 4, 5]
squares = [x ** 2 for x in numbers]
print(squares)  # 출력: [1, 4, 9, 16, 25]
```

```python
monthly_sales = [1200, 1350, 1420, 1500, 1300, 1580, 1620, 1700, 1800, 1850, 1900, 2000]

# 1월, 2월 및 나머지 월 매출액
first, second, *remaining = monthly_sales
print(f'1월 판매액: {first}')
print(f'2월 판매액: {second}')
print(f'나머지 월 판매액: {remaining}')

# 1월, 12월 및 나머지 월 매출액
first, *middle, last = monthly_sales
print(f'첫 달(1월) 판매액: {first}')
print(f'마지막 달(12월) 판매액: {last}')
print(f'중간 달(2~11월)의 판매액: {middle}')
```

## 3-3 함수형 프로그래밍

```python
#람다 함수의 활용
employees = {
    '이지혜': 3000000,
    '구민준': 5000000,
    '방서연': 4100000
}

# 함수 정의: salary * 1.1
adjust_sal = lambda update_sal : update_sal*1.1
# 급여 인상
updated_salaries = {name : adjust_sal(salary)  for name, salary in employees.items()}
## 새로운 딕셔너리 {Key : Value ...}

print(updated_salaries)
```

```python
temperature_celsius = [25.6, 27.8, 30.5, 22.3, 28.9, 31.2, 24.7]

# 함수 정의: 섭씨 -> 화씨
celsius2fahrenheit = lambda c : (c * 9/5) + 32

# 온도 변환
temperature_fahrenheit = list(map(celsius2fahrenheit, temperature_celsius))
print('섭씨 온도 데이터:', temperature_celsius)
print('화씨 온도로 변환:', temperature_fahrenheit)
```

```python
# 함수 정의 및 대문자화
words = ['python', 'data', 'analysis']
uppercase_words = [ word.upper()  for word in words]
print(uppercase_words)
```

```python
employees = [
    {'name': '김지원', 'salary': 3700000},
    {'name': '박민준', 'salary': 7000000},
    {'name': '이서연', 'salary': 4400000},
    {'name': '정현우', 'salary': 3500000}
]

# salary * 1.1
updated_employees = [
    {'name':emp['name'], 'salary':emp['salary']*1.1}
    for emp in employees
]

print('\n급여 인상 후:')

for emp in updated_employees:
    print(f'{emp['name']}: {int(emp['salary']):,}원')
```

```python
numbers = [10, 15, 20, 25, 30]

# 짝수 필터링
even_numbers = list(filter(lambda num:
                      num%2 == 0, numbers))

print(even_numbers)
```

```python
osda_students = [
    {'name': '김지원', 'score': 85},
    {'name': '이민준', 'score': 65},
    {'name': '박서연', 'score': 90},
    {'name': '정현우', 'score': 55},
    {'name': '최예은', 'score': 78}
]

# 70점 이상 학생 목록 생성
passed_students = list(filter(lambda student:
                              student['score'] >= 70, osda_students))
print(f'70점 이상 학생 목록 생성\n {passed_students}')
for student in passed_students:
    print(f'이름: {student["name"]}, 점수: {student["score"]}')
```

```python
from functools import reduce

numbers = [3, 7, 2, 9, 5]

# 최대값 생성
max_value = reduce(lambda x, y : x if (x - y) > 0 else y, numbers)

print(max_value)
```
