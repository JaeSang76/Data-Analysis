###나. str.format() 메서드
####장점

% 방식보다 가독성이 좋다.
{}를 이용하여 삽입 위치를 명확하게 표현할 수 있다.
위치 인자뿐 아니라 키워드 인자를 사용할 수 있어 복잡한 문자열 구성에 유리하다.
숫자 정렬, 소수점 등 다양한 형식 지정이 가능하다.

####단점

문자열과 .format()에 전달되는 변수가 분리되어 있어 f-string보다 코드가 장황해질 수 있다.
단순한 문자열에서는 f-string보다 직관성이 떨어진다.

####적합한 상황

Python 3.6 이전부터 사용되던 코드와의 호환성을 고려할 때
문자열 템플릿을 만들어 두고 여러 값을 나중에 삽입할 때
문자열 형식과 데이터 값을 어느 정도 분리해서 관리하고 싶을 때

```python
name = '홍길동'
age = 30
salary = 3500000
tax_rate = 0.1

# 직원이름, 나이, 월급, 세금, 실수령액 출력
basic_format = print("이름 : {}, 나이 : {}, 월급 : {}, 세금 : {}, 실수령액 : {}".format(name, age, salary, tax_rate, salary - salary*tax_rate))
index_format = print("이름 : {0}, 나이 : {1}, 월급 : {2}, 세금 : {3}, 실수령액 : {4}".format(name, age, salary, tax_rate, salary - salary*tax_rate))
keyword_format = print("이름 : {a}, 나이 : {b}, 월급 : {c}, 세금 : {d}, 실수령액 : {e}".format(a=name, b=age, c=salary, d=tax_rate, e=salary - salary*tax_rate))
```

###다. f-문자열 포맷팅
####장점

변수와 문자열이 한눈에 연결되어 가독성이 가장 좋다.
{} 안에 변수뿐 아니라 표현식을 직접 사용할 수 있다.
코드가 간결하다.
일반적인 Python 프로그램에서 가장 편리하고 직관적이다.

####단점

Python 3.6 이상에서 사용할 수 있다.
문자열을 미리 만들어 두고 나중에 값을 삽입하는 템플릿 용도에는 적합하지 않다.
외부에서 전달되는 문자열을 그대로 f-string의 코드처럼 평가하는 방식은 아니지만, 표현식과 코드가 섞일 수 있으므로 복잡한 로직을 {} 안에 과도하게 넣으면 가독성이 떨어질 수 있다.

####적합한 상황

일반적인 Python 프로그램에서 변수 값을 문자열에 삽입할 때
로그 메시지, 출력문, 파일명 등을 동적으로 생성할 때
현재 작성하는 코드에서 가독성과 생산성을 중요하게 생각할 때

```python
name = '홍길동'
age = 30
salary = 3500000
tax_rate = 0.1

# f-문자열 포매팅
# 이름, 나이, 월급 출력
print(f' 이름 : {name} 나이 : {age}, 월급 : {salary}')
# 직원이름, 나이, 세후 월급 출력
print(f' 이름 : {name} 나이 : {age}, 실수령액 : {salary - salary*tax_rate}')
# 직원이름, 나이, 월급, 세금, 실수령액 추력
print(f' 이름 : {name}, 나이 : {age}, 월급 : {salary}, 세금 : {tax_rate}, 실수령액 : {salary - salary*tax_rate}')
```

```python
# 복수개의 파일(읽기 & 쓰기) 관리
# Reading from a source and splitting into even and odd lines
with (
    open('output.txt', 'r', encoding='utf-8') as src,
    open('even_lines.txt', 'w', encoding='utf-8') as even_file,
    open('odd_lines.txt', 'w', encoding='utf-8') as odd_file
):
    for i, line in enumerate(src, start=1):
        if i % 2 == 0:
            even_file.write(line)
        else:
            odd_file.write(line)

print("Splitting complete!")
```
